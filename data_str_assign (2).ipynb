8. What is a hash table, and how does it relate to dictionaries in Python?
-  A hash table is a data structure that store and retrieve data using a key-value pair. It uses a hash function to compute an index  into an array, from which the desired value can be found. The hash function processes the key and returns an integer that corresponds to an index in the array, where the value associated with the key is stored.
-  In Python, a dictionary  is an implementation of a hash table. When use a dictionary, Python internally uses a hash table to store the key-value pairs. In Python dictionaries, the keys are hashed using a built-in hash function. For a key to be used in a dictionary, it must be hashable, meaning its value must be immutable and have a consistent hash value throughout its lifetime.

14.  How does the “in” keyword work differently for lists and dictionaries?
*   **'in' keyword in list**
    -    The 'in' keyword checks whether a value exists within the list.
    -    It performs a linear search through the list, comparing each element
         until it finds a wanted element or reaches the end of the list.
*   **'in' keyword in Disctionaries**  
    -    The 'in' keyword checks whether a key exists within the dictionary.
    -    In dictionaries, 'in' checks for the presence of a key, not a value.
    -    It uses a hash function to quickly locate the key, making the search
         much faster compared to a list.



```python
my_list = [10, 20, 30, 40]

print(20 in my_list)  # Output: True (20 is in the list)
print(50 in my_list)  # Output: False (50 is not in the list)
```

```python
r_dist = {"a": 1, "b": 2, "c": 3}

# Check for keys
print("b" in r_dist)  # Output: True (key "b" exists)
print(2 in r_dist)    # Output: False (2 is a value, not a key)
```
